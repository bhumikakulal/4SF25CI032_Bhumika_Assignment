**7. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.**


```python
arr = np.random.randint(1, 21, (5, 5))
print("Array:")
print(arr)
border = np.concatenate([
    arr[0, :],      
    arr[1:-1, -1],  
    arr[-1, ::-1],  
    arr[1:-1, 0]    
])
print("Border elements:")
print(border)
```

**8. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.**

```python
arr = np.arange(1, 17).reshape(4, 4)
print(arr)
row_sum = np.sum(arr, axis=1)
column_sum = np.sum(arr, axis=0)
print("Row-wise sum:", row_sum)
print("Column-wise sum:", column_sum)
```

**9. Create a NumPy array of shape (3, 3) with values from 1 to 9. Reshape the array to shape (1, 9) and then to shape (9, 1).**

```python
arr = np.arange(1, 10).reshape(3, 3)
arr1 = arr.reshape(1, 9)
print(arr1)
arr2 = arr.reshape(9, 1)
print(arr2)
```
